# N-Gram Model (Uni-, Bi-, Tri-gram) Generation from Corpus

```python
import nltk
from nltk.tokenize import word_tokenize

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError for punkt_tab

sample_text = "This is a sample sentence for uni-gram generation."

# Tokenize the text into words (uni-grams)
uni_grams = word_tokenize(sample_text.lower()) # Convert to lowercase for consistency

print("Uni-grams:")
print(uni_grams)
```

```python
from nltk.util import ngrams

# Generate bi-grams
bi_grams = list(ngrams(uni_grams, 2))

print("Bi-grams:")
print(bi_grams)
```

```python
# Generate tri-grams
tri_grams = list(ngrams(uni_grams, 3))

print("Tri-grams:")
print(tri_grams)
```
