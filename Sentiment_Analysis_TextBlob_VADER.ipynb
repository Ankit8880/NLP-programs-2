Sentiment Analysis using TextBlob and VADER code

```python
# ============================================================
# SENTIMENT ANALYSIS USING TEXTBLOB AND VADER
# ============================================================

# ------------------------------------------------------------
# 1. Install Required Libraries
# ------------------------------------------------------------

!pip install textblob vaderSentiment -q


# ------------------------------------------------------------
# 2. Import Libraries
# ------------------------------------------------------------

from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import pandas as pd


# ------------------------------------------------------------
# 3. Create Sample Dataset
# ------------------------------------------------------------

texts = [
    "I love this product. It is amazing!",
    "This movie was excellent and very enjoyable.",
    "The food was good and tasty.",
    "I am happy with the service.",
    "This is an average product.",
    "The movie was okay, nothing special.",
    "I don't like this product.",
    "The food was terrible and disappointing.",
    "This movie was very boring.",
    "I hate this service. It is horrible!"
]

df = pd.DataFrame({
    "Text": texts
})

print("Original Dataset:")
display(df)


# ------------------------------------------------------------
# 4. TextBlob Sentiment Analysis
# ------------------------------------------------------------

def textblob_analysis(text):

    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    return polarity, sentiment


df["TextBlob_Polarity"] = df["Text"].apply(
    lambda x: textblob_analysis(x)[0]
)

df["TextBlob_Sentiment"] = df["Text"].apply(
    lambda x: textblob_analysis(x)[1]
)


# ------------------------------------------------------------
# 5. VADER Sentiment Analysis
# ------------------------------------------------------------

analyzer = SentimentIntensityAnalyzer()


def vader_analysis(text):

    scores = analyzer.polarity_scores(text)

    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive"

    elif compound <= -0.05:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    return scores, sentiment


df["VADER_Scores"] = df["Text"].apply(
    lambda x: vader_analysis(x)[0]
)

df["VADER_Sentiment"] = df["Text"].apply(
    lambda x: vader_analysis(x)[1]
)


# ------------------------------------------------------------
# 6. Extract VADER Scores
# ------------------------------------------------------------

df["VADER_Positive"] = df["VADER_Scores"].apply(
    lambda x: x["pos"]
)

df["VADER_Negative"] = df["VADER_Scores"].apply(
    lambda x: x["neg"]
)

df["VADER_Neutral"] = df["VADER_Scores"].apply(
    lambda x: x["neu"]
)

df["VADER_Compound"] = df["VADER_Scores"].apply(
    lambda x: x["compound"]
)


# ------------------------------------------------------------
# 7. Display Complete Results
# ------------------------------------------------------------

result = df[
    [
        "Text",
        "TextBlob_Polarity",
        "TextBlob_Sentiment",
        "VADER_Positive",
        "VADER_Negative",
        "VADER_Neutral",
        "VADER_Compound",
        "VADER_Sentiment"
    ]
]

print("Sentiment Analysis Results:")
display(result)


# ------------------------------------------------------------
# 8. Test Your Own Text
# ------------------------------------------------------------

new_text = "I really love this amazing product!"


# ---------------- TextBlob ----------------

textblob_result = TextBlob(new_text)

textblob_polarity = textblob_result.sentiment.polarity

if textblob_polarity > 0:
    textblob_label = "Positive"

elif textblob_polarity < 0:
    textblob_label = "Negative"

else:
    textblob_label = "Neutral"


# ---------------- VADER ----------------

vader_result = analyzer.polarity_scores(new_text)

vader_compound = vader_result["compound"]

if vader_compound >= 0.05:
    vader_label = "Positive"

elif vader_compound <= -0.05:
    vader_label = "Negative"

else:
    vader_label = "Neutral"


# ------------------------------------------------------------
# 9. Display Custom Text Results
# ------------------------------------------------------------

print("\nCustom Text:")
print(new_text)

print("\nTextBlob Result:")
print("Polarity:", textblob_polarity)
print("Sentiment:", textblob_label)

print("\nVADER Result:")
print("Positive:", vader_result["pos"])
print("Negative:", vader_result["neg"])
print("Neutral:", vader_result["neu"])
print("Compound:", vader_result["compound"])
print("Sentiment:", vader_label)


# ------------------------------------------------------------
# 10. Compare TextBlob and VADER
# ------------------------------------------------------------

comparison = df[
    [
        "Text",
        "TextBlob_Sentiment",
        "VADER_Sentiment"
    ]
]

print("\nTextBlob vs VADER Comparison:")
display(comparison)


# ------------------------------------------------------------
# 11. Count Sentiments
# ------------------------------------------------------------

print("\nTextBlob Sentiment Count:")
print(df["TextBlob_Sentiment"].value_counts())

print("\nVADER Sentiment Count:")
print(df["VADER_Sentiment"].value_counts())
```
