# Text Classification using Naïve Bayes/SVM with TF-IDF

This notebook demonstrates text classification using two popular machine learning algorithms: Naïve Bayes and Support Vector Machines (SVM). We will use Term Frequency-Inverse Document Frequency (TF-IDF) for feature extraction from text data.

```python
# Text Classification using Naive Bayes with TF-IDF

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Dataset
data = {
    "text": [
        "I love this movie, it is amazing",
        "The movie was excellent and wonderful",
        "This film is very good and enjoyable",
        "I really liked this movie",
        "Amazing acting and great story",
        "This is a fantastic movie",

        "I hate this movie",
        "The movie was terrible and boring",
        "This film is very bad",
        "I did not like this movie",
        "Worst movie I have ever seen",
        "The story was boring and disappointing"
    ],

    "label": [
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",
        "positive",

        "negative",
        "negative",
        "negative",
        "negative",
        "negative",
        "negative"
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)

# 2. Separate input and output
X = df["text"]
y = df["label"]

# 3. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# 4. TF-IDF
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training TF-IDF Shape:", X_train_tfidf.shape)
print("Testing TF-IDF Shape:", X_test_tfidf.shape)

# 5. Naive Bayes Model
model = MultinomialNB()

model.fit(X_train_tfidf, y_train)

# 6. Prediction
y_pred = model.predict(X_test_tfidf)

# 7. Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)

# 8. Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 9. Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 10. Test with new sentences
new_text = [
    "This movie was amazing and fantastic",
    "The movie was very boring and terrible"
]

new_text_tfidf = vectorizer.transform(new_text)

predictions = model.predict(new_text_tfidf)

print("\nNew Text Predictions:")

for text, prediction in zip(new_text, predictions):
    print(f"{text} --> {prediction}")
```
