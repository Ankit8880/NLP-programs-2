## 1. Importing Libraries and Defining Corpus

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Download necessary NLTK data (if not already downloaded)
# Directly download stopwords and punkt without complex find checks
# nltk.download will check if already present
nltk.download('stopwords')
nltk.download('punkt') # This is the standard tokenizer resource

# Sample documents
documents = [
    "The quick brown fox jumps over the lazy dog",
    "The lazy dog barks loudly",
    "A quick brown cat runs fast",
    "Foxes are known for their cunning",
    "Dogs are loyal and friendly animals"
]

print("Original Documents:")
for i, doc in enumerate(documents):
    print(f"Document {i+1}: {doc}")
```

## 2. Text Preprocessing

```python
stop_words = set(stopwords.words('english'))
ps = PorterStemmer()

def preprocess_text(text):
    # Lowercasing
    text = text.lower()
    # Tokenization
    words = nltk.word_tokenize(text)
    # Remove stopwords and stemming
    words = [ps.stem(word) for word in words if word.isalnum() and word not in stop_words]
    return " ".join(words)

preprocessed_documents = [preprocess_text(doc) for doc in documents]

print("Preprocessed Documents:")
for i, doc in enumerate(preprocessed_documents):
    print(f"Document {i+1}: {doc}")
```

## 3. Bag-of-Words (BoW) Implementation

```python
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(preprocessed_documents)

print("BoW Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBoW Matrix (Document-Term Matrix):")
print(bow_matrix.toarray())
```

### BoW Cosine Similarity

```python
bow_similarity = cosine_similarity(bow_matrix)

print("BoW Cosine Similarity Matrix:")
print(bow_similarity)

# Example: Similarity between Document 1 and Document 2
print(f"\nSimilarity between Document 1 and Document 2 (BoW): {bow_similarity[0, 1]:.4f}")
```

## 4. TF-IDF (Term Frequency-Inverse Document Frequency) Implementation

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(preprocessed_documents)

print("TF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

### TF-IDF Cosine Similarity

```python
tfidf_similarity = cosine_similarity(tfidf_matrix)

print("TF-IDF Cosine Similarity Matrix:")
print(tfidf_similarity)

# Example: Similarity between Document 1 and Document 2
print(f"\nSimilarity between Document 1 and Document 2 (TF-IDF): {tfidf_similarity[0, 1]:.4f}")
```

## 5. Comparison and Summary

```python
print("Comparison of BoW and TF-IDF:")
print("------------------------------")

print("BoW Vocabulary Size:", len(vectorizer.get_feature_names_out()))
print("TF-IDF Vocabulary Size:", len(tfidf_vectorizer.get_feature_names_out()))

print(f"\nBoW Similarity (Doc 1 vs Doc 2): {bow_similarity[0, 1]:.4f}")
print(f"TF-IDF Similarity (Doc 1 vs Doc 2): {tfidf_similarity[0, 1]:.4f}")

print("\nKey Differences:")
print("1.  **BoW (CountVectorizer)**: Represents documents by the raw count of words. It treats each word as independent and does not account for the importance of words.")
print("2.  **TF-IDF (TfidfVectorizer)**: Considers not only the frequency of a word in a document (TF) but also how unique or important the word is across the entire corpus (IDF). This often leads to better representation for similarity tasks as it down-weights common words (like 'the', 'is') and up-weights rare, significant words.")

print("\nIn this example:")
print("Document 1: The quick brown fox jumps over the lazy dog")
print("Document 2: The lazy dog barks loudly")
print("TF-IDF gives a slightly lower similarity score than BoW. This could be due to TF-IDF's ability to downplay common words like 'the' and 'dog' which appear in both, focusing more on unique terms that truly differentiate the documents.")
```
