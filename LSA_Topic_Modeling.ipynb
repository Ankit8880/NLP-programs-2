```python
# ============================================================
# Topic Modeling using Latent Semantic Analysis (LSA)
# ============================================================

# Step 1: Install Libraries
!pip install scikit-learn pandas matplotlib


# Step 2: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity


# Step 3: Create Sample Dataset
documents = [
    "Artificial intelligence and machine learning are changing technology",
    "Machine learning algorithms are used in artificial intelligence",
    "Deep learning is an important part of machine learning",
    "Football is a popular sport played around the world",
    "Cricket is one of the most popular sports in India",
    "Football and cricket are watched by millions of sports fans",
    "Python and Java are popular programming languages",
    "Java is widely used for software development",
    "Python is commonly used for data science and machine learning"
]

df = pd.DataFrame({
    "Document": documents
})

print("Dataset:")
display(df)


# Step 4: Convert Documents into TF-IDF Matrix
vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(documents)

print("\nTF-IDF Matrix Shape:")
print(tfidf_matrix.shape)


# Step 5: Apply LSA using Truncated SVD
lsa = TruncatedSVD(
    n_components=3,
    random_state=42
)

lsa_matrix = lsa.fit_transform(tfidf_matrix)

print("\nLSA Matrix Shape:")
print(lsa_matrix.shape)


# Step 6: Display the Main Topics
terms = vectorizer.get_feature_names_out()

print("\n========== TOPICS ==========")

for topic_idx, topic in enumerate(lsa.components_):

    top_words = topic.argsort()[-5:][::-1]

    print(f"\nTopic {topic_idx + 1}:")

    for i in top_words:
        print(terms[i], end=" | ")


# Step 7: Display Topic Distribution for Each Document
topic_df = pd.DataFrame(
    lsa_matrix,
    columns=["Topic 1", "Topic 2", "Topic 3"]
)

topic_df["Document"] = documents

print("\n\n========== TOPIC DISTRIBUTION ==========")
display(topic_df)


# Step 8: Assign Dominant Topic to Each Document
topic_df["Dominant Topic"] = lsa_matrix.argmax(axis=1) + 1

print("\n========== DOMINANT TOPIC ==========")
display(topic_df[["Document", "Dominant Topic"]])


# Step 9: Explained Variance
print("\n========== EXPLAINED VARIANCE ==========")

for i, variance in enumerate(lsa.explained_variance_ratio_):
    print(f"Topic {i + 1}: {variance:.4f}")

total_variance = lsa.explained_variance_ratio_.sum()

print(f"\nTotal Explained Variance: {total_variance:.4f}")


# Step 10: Visualize Explained Variance
plt.figure(figsize=(8, 5))

plt.bar(
    ["Topic 1", "Topic 2", "Topic 3"],
    lsa.explained_variance_ratio_
)

plt.xlabel("Topics")
plt.ylabel("Explained Variance")
plt.title("LSA Topic Modeling")

plt.show()


# Step 11: Calculate Document Similarity
similarity = cosine_similarity(lsa_matrix)

similarity_df = pd.DataFrame(
    similarity,
    index=[f"Document {i+1}" for i in range(len(documents))],
    columns=[f"Document {i+1}" for i in range(len(documents))]
)

print("\n========== DOCUMENT SIMILARITY ==========")
display(similarity_df)


# Step 12: Display Most Similar Documents
print("\n========== MOST SIMILAR DOCUMENT PAIRS ==========")

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):

        print(
            f"Document {i+1} & Document {j+1} : "
            f"{similarity[i][j]:.4f}"
        )
```
