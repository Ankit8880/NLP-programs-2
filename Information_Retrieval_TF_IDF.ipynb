```python
# ============================================================
# Information Retrieval System with Ranking using TF-IDF
# ============================================================

# Step 1: Install Required Libraries
!pip install pandas scikit-learn matplotlib


# Step 2: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# STEP 3: CREATE DOCUMENT DATASET
# ============================================================

documents = [
    "Machine learning is a branch of artificial intelligence",
    "Artificial intelligence is used in many modern applications",
    "Deep learning is a type of machine learning",
    "Python is a popular programming language for machine learning",
    "Java is widely used for software development",
    "Data science uses statistics machine learning and programming",
    "Natural language processing helps computers understand human language",
    "Computer vision allows machines to understand images and videos",
    "Web development uses HTML CSS JavaScript and programming",
    "Database management systems are used to store and manage data"
]

df = pd.DataFrame({
    "Document_ID": range(1, len(documents) + 1),
    "Document": documents
})

print("========== DOCUMENT DATASET ==========\n")
display(df)


# ============================================================
# STEP 4: CREATE TF-IDF MATRIX
# ============================================================

vectorizer = TfidfVectorizer(
    stop_words="english"
)

tfidf_matrix = vectorizer.fit_transform(documents)

print("\n========== TF-IDF MATRIX ==========")

print("Number of Documents:", tfidf_matrix.shape[0])
print("Number of Terms:", tfidf_matrix.shape[1])


# ============================================================
# STEP 5: DISPLAY TF-IDF MATRIX
# ============================================================

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\n========== TF-IDF VALUES ==========\n")

display(tfidf_df)


# ============================================================
# STEP 6: CREATE SEARCH FUNCTION
# ============================================================

def search_documents(query):

    # Convert query into TF-IDF vector
    query_vector = vectorizer.transform([query])

    # Calculate cosine similarity
    similarity_scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    # Create result DataFrame
    results = pd.DataFrame({
        "Document_ID": df["Document_ID"],
        "Document": df["Document"],
        "Similarity_Score": similarity_scores
    })

    # Sort documents according to similarity
    results = results.sort_values(
        by="Similarity_Score",
        ascending=False
    )

    return results


# ============================================================
# STEP 7: TEST SEARCH QUERY
# ============================================================

query = "machine learning"

results = search_documents(query)

print("\n==============================================")
print("SEARCH QUERY:", query)
print("==============================================\n")

display(results)


# ============================================================
# STEP 8: SHOW ONLY RELEVANT DOCUMENTS
# ============================================================

relevant_results = results[
    results["Similarity_Score"] > 0
]

print("\n========== RELEVANT DOCUMENTS ==========\n")

display(relevant_results)


# ============================================================
# STEP 9: TOP 5 SEARCH RESULTS
# ============================================================

top_results = results.head(5)

print("\n========== TOP 5 SEARCH RESULTS ==========\n")

display(top_results)


# ============================================================
# STEP 10: VISUALIZE SEARCH RANKING
# ============================================================

plt.figure(figsize=(10, 5))

plt.bar(
    top_results["Document_ID"].astype(str),
    top_results["Similarity_Score"]
)

plt.xlabel("Document ID")
plt.ylabel("Cosine Similarity Score")
plt.title("Document Ranking using TF-IDF")

plt.show()


# ============================================================
# STEP 11: USER INPUT SEARCH
# ============================================================

user_query = input(
    "\nEnter your search query: "
)

user_results = search_documents(user_query)

print("\n==============================================")
print("SEARCH RESULTS FOR:", user_query)
print("==============================================\n")

display(user_results.head(5))


# ============================================================
# STEP 12: DISPLAY RANKED RESULTS
# ============================================================

print("\n========== RANKED DOCUMENTS ==========\n")

for rank, (_, row) in enumerate(
    user_results.head(5).iterrows(),
    start=1
):

    print(f"Rank {rank}")
    print(f"Document ID: {row['Document_ID']}")
    print(f"Similarity Score: {row['Similarity_Score']:.4f}")
    print(f"Document: {row['Document']}")
    print("------------------------------------------")


# ============================================================
# STEP 13: MULTIPLE QUERY EXAMPLES
# ============================================================

queries = [
    "artificial intelligence",
    "python programming",
    "database data",
    "natural language processing",
    "web development"
]

print("\n========== MULTIPLE SEARCH QUERIES ==========\n")

for query in queries:

    result = search_documents(query)

    print(f"\nQuery: {query}")

    for _, row in result.head(3).iterrows():

        print(
            f"Document {row['Document_ID']} "
            f"-> Score: {row['Similarity_Score']:.4f}"
        )


# ============================================================
# STEP 14: FINAL INFORMATION RETRIEVAL FUNCTION
# ============================================================

def information_retrieval(query, top_k=5):

    results = search_documents(query)

    results = results[
        results["Similarity_Score"] > 0
    ]

    return results.head(top_k)


# ============================================================
# STEP 15: FINAL TEST
# ============================================================

query = "deep learning"

final_results = information_retrieval(
    query,
    top_k=5
)

print("\n==============================================")
print("FINAL INFORMATION RETRIEVAL RESULTS")
print("Query:", query)
print("==============================================\n")

display(final_results)
```
