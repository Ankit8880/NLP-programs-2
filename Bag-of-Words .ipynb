## Bag-of-Words (BoW) Vectorization and Representation

**Bag-of-Words (BoW)** is a foundational technique in Natural Language Processing (NLP) for representing text data. The core idea is to transform text into numerical vectors that can be used by machine learning algorithms.

### How it Works:

1.  **Vocabulary Creation:** All unique words from the entire corpus (collection of documents) are extracted to form a vocabulary.
2.  **Word Counting:** For each document, the frequency of each word in the vocabulary is counted.
3.  **Vector Representation:** Each document is then represented as a vector where each dimension corresponds to a word in the vocabulary, and the value in that dimension is the count (or presence/absence) of that word in the document.

### Example Demonstration

Let's demonstrate BoW using a simple corpus and `CountVectorizer` from `sklearn.feature_extraction.text`.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Define a sample corpus
corpus = [
    'The quick brown fox jumps over the lazy dog',
    'A quick brown cat sleeps soundly',
    'The dog is lazy and sleeps a lot'
]

# Initialize CountVectorizer
# max_features can be used to limit the size of the vocabulary
# stop_words='english' removes common English words like 'the', 'a', 'is'
vectorizer = CountVectorizer()

# Fit the vectorizer to the corpus and transform the documents into BoW vectors
X = vectorizer.fit_transform(corpus)

# Get the feature names (i.e., the vocabulary)
feature_names = vectorizer.get_feature_names_out()

# Convert the sparse matrix to a dense array for easier viewing
dense_array = X.toarray()

# Create a DataFrame for better visualization
df_bow = pd.DataFrame(dense_array, columns=feature_names)

print("Original Corpus:")
for i, doc in enumerate(corpus):
    print(f"Document {i+1}: {doc}")

print("\nVocabulary:", feature_names)

print("\nBag-of-Words Representation (Document-Term Matrix):")
display(df_bow)
```
