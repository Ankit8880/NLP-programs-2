```python
# ============================================================
# Information Extraction (IE) from
# Structured / Unstructured Documents
# ============================================================

# Step 1: Install Required Libraries
!pip install pandas spacy openpyxl

# Download English NLP Model
!python -m spacy download en_core_web_sm


# Step 2: Import Libraries
import pandas as pd
import spacy
import re
from IPython.display import display


# Step 3: Load spaCy Model
nlp = spacy.load("en_core_web_sm")


# ============================================================
# PART A: INFORMATION EXTRACTION FROM UNSTRUCTURED TEXT
# ============================================================

# Step 4: Create Unstructured Document
text = """
John Smith is working at Google India as a Software Engineer.
He is located in Bangalore, India.

His email address is john.smith@gmail.com and his phone number
is +91-9876543210.

John has experience in Python, Java, Machine Learning,
Data Science and SQL.

He joined Google on 15 January 2024.
The project deadline is 30 September 2026.
"""


print("========== ORIGINAL DOCUMENT ==========\n")
print(text)


# Step 5: Process Text using spaCy
doc = nlp(text)


# Step 6: Named Entity Recognition (NER)
print("\n========== NAMED ENTITIES ==========\n")

entities = []

for ent in doc.ents:
    entities.append({
        "Entity": ent.text,
        "Label": ent.label_,
        "Description": spacy.explain(ent.label_)
    })

entity_df = pd.DataFrame(entities)

display(entity_df)


# Step 7: Extract Person Names
persons = [
    ent.text
    for ent in doc.ents
    if ent.label_ == "PERSON"
]

print("\n========== PERSON ==========")
print(persons)


# Step 8: Extract Organizations
organizations = [
    ent.text
    for ent in doc.ents
    if ent.label_ == "ORG"
]

print("\n========== ORGANIZATION ==========")
print(organizations)


# Step 9: Extract Locations
locations = [
    ent.text
    for ent in doc.ents
    if ent.label_ in ["GPE", "LOC"]
]

print("\n========== LOCATION ==========")
print(locations)


# Step 10: Extract Dates
dates = [
    ent.text
    for ent in doc.ents
    if ent.label_ == "DATE"
]

print("\n========== DATES ==========")
print(dates)


# ============================================================
# PART B: REGULAR EXPRESSION BASED INFORMATION EXTRACTION
# ============================================================

# Step 11: Extract Email Address
emails = re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    text
)

print("\n========== EMAIL ==========")
print(emails)


# Step 12: Extract Phone Number
phones = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    text
)

print("\n========== PHONE NUMBER ==========")
print(phones)


# Step 13: Extract Skills
skills_list = [
    "Python",
    "Java",
    "Machine Learning",
    "Data Science",
    "SQL",
    "C++",
    "JavaScript",
    "HTML",
    "CSS",
    "Deep Learning"
]

found_skills = []

for skill in skills_list:
    if re.search(
        r'\b' + re.escape(skill) + r'\b',
        text,
        re.IGNORECASE
    ):
        found_skills.append(skill)

print("\n========== SKILLS ==========")
print(found_skills)


# ============================================================
# PART C: CREATE FINAL EXTRACTED INFORMATION TABLE
# ============================================================

# Step 14: Combine Extracted Information

person_name = persons[0] if persons else None
organization = organizations[0] if organizations else None
location = locations[0] if locations else None

extracted_information = {
    "Name": person_name,
    "Organization": organization,
    "Location": location,
    "Email": emails[0] if emails else None,
    "Phone": phones[0] if phones else None,
    "Dates": ", ".join(dates),
    "Skills": ", ".join(found_skills)
}

final_df = pd.DataFrame([extracted_information])


print("\n========== EXTRACTED INFORMATION ==========\n")

display(final_df)


# ============================================================
# PART D: INFORMATION EXTRACTION FROM STRUCTURED DATA
# ============================================================

# Step 15: Create Structured Dataset

structured_data = {
    "Name": [
        "Rahul",
        "Priya",
        "Aman",
        "Neha"
    ],

    "Email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "aman@gmail.com",
        "neha@gmail.com"
    ],

    "Department": [
        "CSE",
        "AI",
        "IT",
        "CSE"
    ],

    "Skills": [
        "Python, SQL",
        "Machine Learning, Python",
        "Java, SQL",
        "Python, Java"
    ]
}

structured_df = pd.DataFrame(structured_data)


print("\n========== STRUCTURED DATA ==========\n")

display(structured_df)


# Step 16: Extract Specific Information
print("\n========== CSE STUDENTS ==========\n")

cse_students = structured_df[
    structured_df["Department"] == "CSE"
]

display(cse_students)


# ============================================================
# PART E: EXTRACT INFORMATION FROM CSV FILE
# ============================================================

# Step 17: Save Structured Data as CSV

structured_df.to_csv(
    "student_information.csv",
    index=False
)

print("\nCSV file created successfully!")


# Step 18: Read CSV File

loaded_df = pd.read_csv(
    "student_information.csv"
)

print("\n========== DATA LOADED FROM CSV ==========\n")

display(loaded_df)


# ============================================================
# PART F: SEARCH INFORMATION FROM DOCUMENT
# ============================================================

# Step 19: Search for Specific Information

search_words = [
    "Python",
    "Google",
    "Machine Learning",
    "Bangalore"
]

print("\n========== SEARCH RESULTS ==========\n")

for word in search_words:

    if word.lower() in text.lower():
        print(f"{word}: Found")
    else:
        print(f"{word}: Not Found")


# ============================================================
# PART G: FINAL SUMMARY
# ============================================================

print("\n==============================================")
print("        INFORMATION EXTRACTION SUMMARY")
print("==============================================")

print("Name:", person_name)
print("Organization:", organization)
print("Location:", location)
print("Email:", emails)
print("Phone:", phones)
print("Dates:", dates)
print("Skills:", found_skills)

print("\nInformation Extraction Completed Successfully!")
```
