## Prepare the Corpus

First, we need to define our custom corpus and preprocess it. This involves tokenization, lowercasing, and removing punctuation.

```python
import nltk
from nltk.tokenize import word_tokenize
import re

# Download necessary NLTK data (if not already downloaded)
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

# Sample custom corpus
corpus = [
    "The quick brown fox jumps over the lazy dog.",
    "I love natural language processing and machine learning.",
    "Word embeddings are a powerful tool in NLP.",
    "Gensim is a popular library for topic modeling and word embeddings.",
    "Artificial intelligence is transforming the world."
]

def preprocess_text(text):
    text = text.lower() # Convert to lowercase
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    tokens = word_tokenize(text) # Tokenize the text
    return tokens

# Preprocess the corpus
processed_corpus = [preprocess_text(sentence) for sentence in corpus]

print("Original Corpus:")
for sentence in corpus:
    print(f"- {sentence}")

print("\nProcessed Corpus (tokens):")
for tokens in processed_corpus:
    print(f"- {tokens}")
```

## Train the Word2Vec Model

Now we will use Gensim's `Word2Vec` model to train word embeddings on our preprocessed corpus. We'll set some common parameters like `vector_size`, `window`, `min_count`, and `workers`.

```python
from gensim.models import Word2Vec

# Initialize and train the Word2Vec model
# vector_size: Dimensionality of the word vectors.
# window: Maximum distance between the current and predicted word within a sentence.
# min_count: Ignores all words with total frequency lower than this.
# workers: Use these many worker threads to train the model.

model = Word2Vec(sentences=processed_corpus, vector_size=100, window=5, min_count=1, workers=4)

# Build the vocabulary
model.build_vocab(processed_corpus)

# Train the model
model.train(processed_corpus, total_examples=model.corpus_count, epochs=10)

print("Word2Vec model trained successfully!")
print(f"Vocabulary size: {len(model.wv)}")
```

## Explore Word Embeddings

Let's see some results from our trained Word2Vec model. We can find words similar to a given word and retrieve the vector for a specific word.

```python
# Find words similar to 'nlp'
print("\nWords similar to 'nlp':")
if 'nlp' in model.wv:
    similar_words = model.wv.most_similar('nlp')
    for word, score in similar_words:
        print(f"- {word}: {score:.4f}")
else:
    print(" 'nlp' not in vocabulary.")

# Find words similar to 'machine'
print("\nWords similar to 'machine':")
if 'machine' in model.wv:
    similar_words = model.wv.most_similar('machine')
    for word, score in similar_words:
        print(f"- {word}: {score:.4f}")
else:
    print(" 'machine' not in vocabulary.")


# Get the vector for 'fox'
print("\nVector for 'fox':")
if 'fox' in model.wv:
    fox_vector = model.wv['fox']
    print(fox_vector)
else:
    print(" 'fox' not in vocabulary.")
```
