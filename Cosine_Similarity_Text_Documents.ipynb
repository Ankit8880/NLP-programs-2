# Cosine Similarity Computation between Text Documents

### What is Cosine Similarity?

Cosine similarity is a metric used to measure how similar two non-zero vectors are. It measures the cosine of the angle between two vectors. The closer the cosine value is to 1, the smaller the angle and the higher the similarity between the vectors. A value of 0 indicates orthogonality (no similarity), and a value of -1 indicates complete opposition.

### How is it applied to Text?

When applied to text, documents are first converted into numerical vectors. This process is called **vectorization**. Common vectorization techniques include:

*   **Bag-of-Words (BoW)**: Represents a text document as a bag (multiset) of its words, disregarding grammar and even word order, but keeping multiplicity. Each dimension of the vector corresponds to a word in the vocabulary, and the value is the frequency of that word in the document.
*   **TF-IDF (Term Frequency-Inverse Document Frequency)**: A statistical measure that evaluates how relevant a word is to a document in a collection of documents. It increases proportionally to the number of times a word appears in the document but is offset by the frequency of the word in the corpus.
*   **Word Embeddings (e.g., Word2Vec, GloVe, FastText)**: Represent words as dense vectors in a continuous vector space, where words with similar meanings are located closer to each other.
*   **Sentence/Document Embeddings (e.g., Doc2Vec, Universal Sentence Encoder, BERT embeddings)**: Extend word embeddings to entire sentences or documents.

Once documents are vectorized, cosine similarity can be calculated between their respective vectors. The formula for cosine similarity between two vectors A and B is:

$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \cdot \|B\|} = \frac{\sum_{i=1}^{n} A_i B_i}{\sqrt{\sum_{i=1}^{n} A_i^2} \sqrt{\sum_{i=1}^{n} B_i^2}} $$

Where:
*   $A_i$ and $B_i$ are components of vector A and B respectively.
*   $A \cdot B$ is the dot product of the vectors.
*   $\|A\|$ and $\|B\|$ are the Euclidean magnitudes (L2 norms) of the vectors.

### Why is it useful for Text Analysis?

*   **Robust to Document Length**: Cosine similarity is particularly effective for text data because it is not influenced by the length of the documents. If two documents are on the same topic but one is much longer, their Euclidean distance might be large, but their cosine similarity will remain high because their orientation in the vector space is similar.
*   **Handles High Dimensionality**: Text vectorization often results in very high-dimensional sparse vectors. Cosine similarity performs well in such spaces.
*   **Applications**: It's widely used in information retrieval (e.g., finding relevant documents), natural language processing (e.g., text summarization, plagiarism detection, chatbot responses), and recommender systems (e.g., suggesting similar articles).

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "I love machine learning and artificial intelligence",
    "I love artificial intelligence and machine learning",
    "Python is useful for data science",
    "I enjoy playing cricket and football"
]

print("Original Documents:\n")

for i, doc in enumerate(documents):
    print(f"Document {i + 1}: {doc}")

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("\nTF-IDF Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
display(pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"Document {i + 1}" for i in range(len(documents))]
))

similarity_matrix = cosine_similarity(tfidf_matrix)

print("\nCosine Similarity Matrix:")

df_similarity = pd.DataFrame(
    similarity_matrix,
    index=[f"Document {i + 1}" for i in range(len(documents))],
    columns=[f"Document {i + 1}" for i in range(len(documents))]
)

display(df_similarity)

print("\nSimilarity between Documents:")

for i in range(len(documents)):
    for j in range(i + 1, len(documents)):
        print(
            f"Document {i + 1} and Document {j + 1}: "
            f"{similarity_matrix[i][j]:.4f}"
        )
```
